```python
import pandas as pd
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
print("IRIS Dataset:")
print(df.head())
```

```python
print("\nNull Values Before Filling:")
print(df.isnull().sum())
```

```python
df.fillna(df.mean(), inplace=True)
```

```python
print("\nNull Values After Filling:")
print(df.isnull().sum())
```

```python
print("\nMean:")
print(df.mean())
```

```python
print("\nMedian:")
print(df.median())
```

```python
print("\nMode:")
print(df.mode())
```
